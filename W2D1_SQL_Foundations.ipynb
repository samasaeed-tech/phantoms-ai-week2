```python
import sqlite3
conn = sqlite3.connect('store.db')
cursor = conn.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS products (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    category TEXT NOT NULL,
    price REAL NOT NULL
)
''')

products_data = [
    ('Laptop Dell', 'Electronics', 850.0),
    ('Wireless Mouse', 'Electronics', 25.5),
    ('Coffee Maker', 'Home', 120.0),
    ('Keyboard', 'Electronics', 60.0),
    ('Air Fryer', 'Home', 180.0)
]

cursor.executemany('''
INSERT INTO products (name, category, price)
VALUES (?, ?, ?)
''', products_data)

conn.commit()

print("--- 1. منتجات قسم الـ Electronics فقط ---")
cursor.execute("SELECT * FROM products WHERE category = 'Electronics'")
for row in cursor.fetchall():
    print(row)

print("\n--- 2. المنتجات متأثرة من الأعلى سعراً للأقل ---")
cursor.execute("SELECT * FROM products ORDER BY price DESC")
for row in cursor.fetchall():
    print(row)

conn.close()
```
