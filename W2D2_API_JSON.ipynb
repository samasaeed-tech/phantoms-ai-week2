```python
import sqlite3
import requests

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)

print(f"Status Code: {response.status_code}")

if response.status_code == 200:
    data = response.json()
    conn = sqlite3.connect('store.db')
    cursor = conn.cursor()

    cursor.execute('''
    CREATE TABLE IF NOT EXISTS products (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        category TEXT NOT NULL,
        price REAL NOT NULL
    )
    ''')

    extracted_data = []
    for item in data[:5]:
        item_id = item['id']
        name = item['name']
        category = item['username']  # الحقل الثاني
        price = round(len(name) * 12.5, 2)  # الحقل الثالث (حساب سعر افتراضي)

        extracted_data.append((item_id, name, category, price))

    cursor.executemany('''
    INSERT OR REPLACE INTO products (id, name, category, price)
    VALUES (?, ?, ?, ?)
    ''', extracted_data)

    conn.commit()
    print("تم جلب البيانات وإدراجها بنجاح في قاعدة البيانات!\n")
    print("--- 1. استعلام باستخدام WHERE (البحث برقم ID محدد) ---")
    cursor.execute("SELECT * FROM products WHERE id = 1")
    for row in cursor.fetchall():
        print(f"ID: {row[0]} | Name: {row[1]} | Category: {row[2]} | Price: ${row[3]}")

    print("\n--------------------------------------------------\n")

    print("--- 2. استعلام باستخدام ORDER BY (ترتيب تنازلي حسب السعر) ---")
    cursor.execute("SELECT * FROM products ORDER BY price DESC")
    for row in cursor.fetchall():
        print(f"ID: {row[0]} | Name: {row[1]} | Category: {row[2]} | Price: ${row[3]}")

    conn.close()
else:
    print("فشل الاتصال بالـ API!")
```
